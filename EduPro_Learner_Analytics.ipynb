# Phase 2: Exploratory Data Analysis (EDA)

###02.1 – Learner Age Group Distribution

In this step, we will analyze the age distribution of learners on the EduPro platform.

Learners will be categorized into the following age groups:

- Below 18
- 18–25
- 26–35
- 36–45
- Above 45

We will calculate the number and percentage of unique learners in each age group to identify the major learner segments.

This analysis will help us understand which age groups are most actively represented on the platform.

```python
# Creating age groups for learners

age_bins = [0, 17, 25, 35, 45, float("inf")]

age_labels = [
    "Below 18",
    "18–25",
    "26–35",
    "36–45",
    "Above 45"
]

# Assigning age groups in the integrated dataset
edupro_df["Age Group"] = pd.cut(
    edupro_df["Age"],
    bins=age_bins,
    labels=age_labels,
    right=True
)

# Counting unique learners in each age group
learner_age_df = (
    edupro_df[["UserID", "Age Group"]]
    .drop_duplicates(subset="UserID")
)

age_group_summary = (
    learner_age_df["Age Group"]
    .value_counts(sort=False)
    .rename_axis("Age Group")
    .reset_index(name="Unique Learners")
)

# Calculating percentage distribution
age_group_summary["Percentage"] = (
    age_group_summary["Unique Learners"]
    / age_group_summary["Unique Learners"].sum()
) * 100

display(age_group_summary.round(2))
```

### 02.2 Gender-wise Learner Distribution

In this step, we will analyze the gender distribution of learners on the EduPro platform.

We will calculate the number and percentage of unique learners belonging to each gender category.

This analysis will help us understand gender participation on the platform and identify the distribution of learners across different gender groups.

```python
# Analyzing gender distribution among unique learners

learner_gender_df = (
    edupro_df[["UserID", "Gender"]]
    .drop_duplicates(subset="UserID")
)

gender_summary = (
    learner_gender_df["Gender"]
    .value_counts()
    .reset_index()
)

gender_summary.columns = ["Gender", "Unique Learners"]

# Calculating percentage distribution
gender_summary["Percentage"] = (
    gender_summary["Unique Learners"]
    / gender_summary["Unique Learners"].sum()
) * 100

display(gender_summary.round(2))
```

### 02.3 Course Category-wise Enrollment Distribution

In this step, we will analyze the distribution of course enrollments across different course categories on the EduPro platform.

We will calculate the total number of enrollments and their percentage contribution for each course category.

This analysis will help us identify the most and least preferred course categories among learners and understand the overall distribution of learning interests.

```python
# Analyzing enrollment distribution across course categories

category_summary = (
    edupro_df["CourseCategory"]
    .value_counts()
    .reset_index()
)

category_summary.columns = ["Course Category", "Total Enrollments"]

# Calculating percentage contribution
category_summary["Percentage"] = (
    category_summary["Total Enrollments"]
    / category_summary["Total Enrollments"].sum()
) * 100

# Sorting categories by enrollment count
category_summary = category_summary.sort_values(
    by="Total Enrollments",
    ascending=False
).reset_index(drop=True)

display(category_summary.round(2))
```

### 02.4 Course Type Enrollment Analysis

In this step, we will analyze the distribution of enrollments across different course types.

This analysis will help us understand learner participation in different course types, such as Free and Paid courses, and identify which type receives more enrollments.

```python
# Calculating total enrollments by course type
course_type_summary = (
    edupro_df.groupby("CourseType")
    .size()
    .reset_index(name="Total Enrollments")
)

# Calculating percentage distribution
course_type_summary["Percentage"] = (
    course_type_summary["Total Enrollments"]
    / course_type_summary["Total Enrollments"].sum()
) * 100

# Sorting by total enrollments
course_type_summary = course_type_summary.sort_values(
    by="Total Enrollments",
    ascending=False
).reset_index(drop=True)

display(course_type_summary.round(2))
```

### 02.5 Analyzing Course Level Distribution

In this step, we will analyze the distribution of enrollments across different course levels, including Beginner, Intermediate, and Advanced.

This analysis will help us understand which learning levels are more popular among learners and identify the overall preference for different levels of course difficulty.

```python
# Calculating total enrollments by course level
course_level_summary = (
    edupro_df["CourseLevel"]
    .value_counts()
    .reset_index()
)

course_level_summary.columns = ["Course Level", "Total Enrollments"]

# Calculating percentage distribution
course_level_summary["Percentage"] = (
    course_level_summary["Total Enrollments"]
    / course_level_summary["Total Enrollments"].sum()
) * 100

# Sorting by total enrollments
course_level_summary = course_level_summary.sort_values(
    by="Total Enrollments",
    ascending=False
).reset_index(drop=True)

display(course_level_summary.round(2))
```

### 02.6 Analyzing Overall Learner Enrollment Activity

In this step, we will analyze how frequently learners enroll in courses on the EduPro platform.

We will calculate the number of enrollments per learner and examine the distribution of enrollment activity. This will help us understand whether learners generally enroll in a small number of courses or whether a smaller group of learners accounts for multiple enrollments.

```python
# Calculating the number of enrollments per learner
learner_enrollment_summary = (
    edupro_df.groupby("UserID")
    .size()
    .reset_index(name="Total Enrollments")
)

# Calculating overall enrollment activity statistics
enrollment_activity_summary = learner_enrollment_summary[
    "Total Enrollments"
].describe().to_frame(name="Value")

# Adding the average number of enrollments per learner
enrollment_activity_summary.loc[
    "Average Enrollments per Learner", "Value"
] = learner_enrollment_summary["Total Enrollments"].mean()

# Displaying the summary
display(enrollment_activity_summary.round(2))
```

### 02.7 Analyzing Learner Enrollment Frequency

In this step, we will examine how many learners have enrolled in 1, 2, 3, or more courses.

This analysis will help us understand the distribution of learner activity and identify the proportion of learners with different enrollment frequencies.

```python
# Calculating the frequency of enrollments per learner
enrollment_frequency = (
    learner_enrollment_summary["Total Enrollments"]
    .value_counts()
    .sort_index()
    .reset_index()
)

enrollment_frequency.columns = [
    "Number of Enrollments",
    "Number of Learners"
]

# Calculating percentage of learners
enrollment_frequency["Percentage"] = (
    enrollment_frequency["Number of Learners"]
    / enrollment_frequency["Number of Learners"].sum()
) * 100

# Displaying the enrollment frequency distribution
display(enrollment_frequency.round(2))
```

### 02.8 Analyzing Course Enrollment Across Age Groups

In this step, we will examine how course enrollments are distributed across different learner age groups.

By combining age group information with course enrollment records, we will identify which age groups contribute the most to different course categories.

This analysis will help us understand the learning interests and course preferences of different age segments.

```python
# Analyzing course category preferences across age groups

age_category_summary = (
    edupro_df.groupby(["Age Group", "CourseCategory"], observed=False)
    .size()
    .reset_index(name="Total Enrollments")
)

# Creating a pivot table for age group and course category distribution

age_category_pivot = age_category_summary.pivot(
    index="Age Group",
    columns="CourseCategory",
    values="Total Enrollments"
).fillna(0)

# Displaying the age group and course category distribution

display(age_category_pivot.astype(int))
```

### 02.9 Visualizing Age Group vs Course Category Distribution

In this step, we will create a heatmap to visualize the relationship between learner age groups and course category preferences.

The heatmap will help identify which course categories receive the highest enrollments across different age groups and highlight variations in learning interests among demographic segments.

```python
# Visualizing age group and course category distribution using a heatmap

plt.figure(figsize=(16, 7))

sns.heatmap(
    age_category_pivot,
    annot=True,
    fmt="d",
    cmap="YlGnBu",
    linewidths=0.5,
    linecolor="white",
    cbar_kws={"label": "Total Enrollments"}
)

plt.title(
    "Course Category Preferences Across Age Groups",
    fontsize=16,
    fontweight="bold",
    pad=18
)

plt.xlabel("Course Category", fontsize=12, fontweight="bold")
plt.ylabel("Age Group", fontsize=12, fontweight="bold")

plt.xticks(rotation=35, ha="right")
plt.yticks(rotation=0)

plt.tight_layout()
plt.show()
```

### 02.10 Analyzing Course Level Preferences Across Genders

In this step, we will examine how learners of different genders enroll in courses across Beginner, Intermediate, and Advanced levels.

This analysis will help identify differences in course level preferences between male and female learners and provide insights into learning patterns across gender groups.

```python
# Analyzing course level preferences across genders

gender_level_summary = (
    edupro_df.groupby(["Gender", "CourseLevel"])
    .size()
    .reset_index(name="Total Enrollments")
)

# Creating a pivot table for gender and course level distribution

gender_level_pivot = gender_level_summary.pivot(
    index="Gender",
    columns="CourseLevel",
    values="Total Enrollments"
).fillna(0)

# Displaying the gender-wise course level distribution

display(gender_level_pivot.astype(int))
```

### 02.11 Visualizing Gender-wise Course Level Preferences

In this step, we will create a grouped bar chart to compare course level enrollments between female and male learners.

This visualization will help us understand how course level preferences vary across gender groups and identify similarities or differences in their enrollment patterns.

#Phase 3 - Behavioral Analysis

### 03.1 Enrollment Concentration Analysis

This analysis examines how total course enrollments are distributed among learners. It identifies whether enrollment activity is evenly distributed or concentrated among a smaller group of highly active learners.

```python
# Calculating enrollment concentration among learners

learner_activity = learner_enrollment_summary[
    ["Total Enrollments"]
].copy()

learner_activity = learner_activity.sort_values(
    by="Total Enrollments",
    ascending=False
).reset_index(drop=True)

total_learners = len(learner_activity)
total_enrollments = learner_activity["Total Enrollments"].sum()

top_10_count = max(1, int(total_learners * 0.10))
top_20_count = max(1, int(total_learners * 0.20))

top_10_enrollments = learner_activity.head(top_10_count)[
    "Total Enrollments"
].sum()

top_20_enrollments = learner_activity.head(top_20_count)[
    "Total Enrollments"
].sum()

concentration_summary = pd.DataFrame({
    "Metric": [
        "Total Learners",
        "Total Enrollments",
        "Top 10% Learners' Enrollment Share (%)",
        "Top 20% Learners' Enrollment Share (%)"
    ],
    "Value": [
        total_learners,
        total_enrollments,
        round((top_10_enrollments / total_enrollments) * 100, 2),
        round((top_20_enrollments / total_enrollments) * 100, 2)
    ]
})

display(concentration_summary)
```

### 03.2 Beginner vs Advanced Course Enrollment Behavior

This analysis examines enrollment patterns across different course levels to understand learner preferences for Beginner, Intermediate, and Advanced courses.

**Note:** A learner may enroll in courses across multiple course levels. Therefore, the learner percentages calculated for Beginner, Intermediate, and Advanced levels may overlap and are not expected to add up to 100%. These percentages represent the share of total unique learners enrolled in each course level.

###03.2.1 Unique Learners Enrolled Aross Course Levels

```python
# Analyzing unique learners enrolled in each course level

learner_level_summary = (
    edupro_df.groupby("CourseLevel")["UserID"]
    .nunique()
    .reset_index(name="Unique Learners")
)

# Calculating percentage of total learners
total_unique_learners = edupro_df["UserID"].nunique()

learner_level_summary["Share of Total Learners (%)"] = (
    learner_level_summary["Unique Learners"] /
    total_unique_learners
) * 100

# Sorting by number of unique learners
learner_level_summary = learner_level_summary.sort_values(
    by="Unique Learners",
    ascending=False
).reset_index(drop=True)

display(learner_level_summary.round(2))
```

###03.2.2 Average Enrollments per Learner Across Course Levels

This analysis calculates the average number of course enrollments per unique learner for each course level. It helps compare enrollment frequency across Beginner, Intermediate, and Advanced courses.

```python
# Calculating average enrollments per learner across course levels

level_engagement_summary = (
    edupro_df.groupby("CourseLevel")
    .agg(
        Total_Enrollments=("TransactionID", "count"),
        Unique_Learners=("UserID", "nunique")
    )
    .reset_index()
)

# Calculating average enrollments per learner
level_engagement_summary["Avg_Enrollments_Per_Learner"] = (
    level_engagement_summary["Total_Enrollments"] /
    level_engagement_summary["Unique_Learners"]
)

# Sorting by average enrollment frequency
level_engagement_summary = level_engagement_summary.sort_values(
    by="Avg_Enrollments_Per_Learner",
    ascending=False
).reset_index(drop=True)

display(level_engagement_summary.round(2))
```

##03.3 Learner Engagement Segmentation

This analysis segments active learners into Low, Moderate, and High engagement groups based on their total course enrollments. It helps identify how learner participation is distributed across different engagement levels.

For this analysis:

- Low Engagement: 1–2 enrollments
- Moderate Engagement: 3–4 enrollments
- High Engagement: 5 or more enrollments

These are analytical categories created for this project and are not predefined platform classifications.

```python
# Calculating total enrollments for each active learner

learner_enrollment_count = (
    edupro_df.groupby("UserID")
    .size()
    .reset_index(name="Total Enrollments")
)

# Creating engagement segments
learner_enrollment_count["Engagement Segment"] = pd.cut(
    learner_enrollment_count["Total Enrollments"],
    bins=[0, 2, 4, float("inf")],
    labels=["Low Engagement", "Moderate Engagement", "High Engagement"]
)

# Summarizing learner engagement segments
engagement_summary = (
    learner_enrollment_count.groupby(
        "Engagement Segment",
        observed=False
    )
    .agg(
        Number_of_Learners=("UserID", "count")
    )
    .reset_index()
)

# Calculating percentage of active learners
total_active_learners = learner_enrollment_count["UserID"].nunique()

engagement_summary["Percentage"] = (
    engagement_summary["Number_of_Learners"] /
    total_active_learners
) * 100

# Displaying the final engagement summary
display(engagement_summary.round(2))
```

### 04.5 — Age Group vs Course Level Analysis

This step examines how course level preferences vary across different learner age groups.

A cross-tabulation analysis will compare Beginner, Intermediate, and Advanced course enrollments across age groups.

This analysis helps identify differences in learning-level preferences among demographic segments.

```python
# Age Group vs Course Level Enrollment Analysis

age_level_summary = pd.crosstab(
    edupro_df["Age Group"],
    edupro_df["CourseLevel"]
)

# Ensure all course levels and age groups are represented
age_level_summary = age_level_summary.reindex(
    index=["Below 18", "18–25", "26–35", "36–45", "Above 45"],
    columns=["Beginner", "Intermediate", "Advanced"],
    fill_value=0
)

print("ENROLLMENTS BY AGE GROUP AND COURSE LEVEL")
display(age_level_summary)

# Percentage distribution within each age group
age_level_percentage = (
    age_level_summary
    .div(age_level_summary.sum(axis=1).replace(0, 1), axis=0)
    * 100
)

print("\nCOURSE LEVEL PREFERENCE (%) WITHIN EACH AGE GROUP")
display(age_level_percentage.round(2))
```

```python
# Course Type vs Course Level Enrollment Analysis

course_type_level_summary = pd.crosstab(
    edupro_df["CourseType"],
    edupro_df["CourseLevel"]
)

print("ENROLLMENTS BY COURSE TYPE AND COURSE LEVEL")
display(course_type_level_summary)

# Percentage distribution within each course type
course_type_level_percentage = (
    course_type_level_summary
    .div(
        course_type_level_summary.sum(axis=1).replace(0, 1),
        axis=0
    ) * 100
)

print("\nCOURSE LEVEL PREFERENCE (%) WITHIN EACH COURSE TYPE")
display(course_type_level_percentage.round(2))
```
